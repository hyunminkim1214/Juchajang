```python
#파일 저장 형식
import urllib.request
import json

def get_request_url(url):
    req = urllib.request.Request(url)
    try:
        response = urllib.request.urlopen(req)
        if response.getcode() == 200:
            print(f"[OSM: 성공] {url} 접속 완료")
            return response.read().decode('utf-8')
        else:
            print(f"[OSM: 에러] {url} 접속 실패 - {response.getcode()}")
            return None
    except Exception as e:
        print(f"[OSM: 에러] {e}")
        return None

def save_to_json(data, filename):
    with open(filename, 'w', encoding='utf-8') as outfile:
        json.dump(data, outfile, indent=4, ensure_ascii=False)

# API 요청 URL (예시)

# API 요청 및 응답 받기
response_data = get_request_url(url)

if response_data:
    # JSON 파싱
    try:
        json_data = json.loads(response_data)
    except json.JSONDecodeError as e:
        print(f"[OSM: 에러] JSON 파싱 실패: {e}")
        json_data = None

    # 파일 저장
    if json_data:
        save_to_json(json_data, "output.json")
        print("output.json 파일이 생성되었습니다.")
```

```python
with open('output.json') as f:
    js = json.loads(f.read())
df = pd.DataFrame(js)
```

```python
# 1. 'row' 키의 값(딕셔너리 리스트)만 추출
parking_info_rows = json_data['GetParkingInfo']['row']

# 2. 추출된 리스트를 pd.DataFrame() 함수에 전달하여 DataFrame 생성
df = pd.DataFrame(parking_info_rows)

# 3. 결과 확인
display(df)
```

```python
df['NOW_PRK_VHCL_UPDT_TM'] = pd.to_datetime(df['NOW_PRK_VHCL_UPDT_TM'])
df = df.set_index('NOW_PRK_VHCL_UPDT_TM')
```

```python
display(df)
df['hour'] = df.index.hour
df['dayofweek'] = df.index.dayofweek
df['is_weekend'] = df['dayofweek'].isin([5, 6]).astype(int)
df['NOW_PRK_VHCL_CNT_lag_1h'] = df.groupby('PKLT_CD')['NOW_PRK_VHCL_CNT'].shift(1)
```

```python
import pandas as pd
import numpy as np
import json
from prophet import Prophet
import matplotlib.pyplot as plt



# 예측 결과를 저장할 딕셔너리
forecasts = {}

# 전체 주차장의 고유한 코드 리스트
unique_parking_lots = df['PKLT_CD'].unique()

print("--- 각 주차장별 예측 시작 ---")
for pklt_cd in unique_parking_lots:

    # 2-1. 현재 주차장의 데이터만 필터링
    df_single_lot = df[df['PKLT_CD'] == pklt_cd]

    # 2-2. Prophet 모델 생성 및 학습
    m = Prophet(
        weekly_seasonality=True,
        daily_seasonality=True,
        yearly_seasonality=False # 데이터가 짧으므로 비활성화
    )
    m.fit(df_single_lot)

    # 2-3. 향후 24시간 예측
    future = m.make_future_dataframe(periods=24, freq='H')
    forecast = m.predict(future)

    # 2-4. 예측 결과에 주차장 정보 추가
    forecast['PKLT_CD'] = pklt_cd
    forecasts[pklt_cd] = forecast

    print(f"- {pklt_cd} 주차장 예측 완료")

print("--- 모든 주차장 예측 완료 ---")


all_forecasts_df = pd.concat(forecasts.values(), ignore_index=True)


print(all_forecasts_df[['ds', 'PKLT_CD', 'yhat', 'yhat_lower', 'yhat_upper']].tail())
```
